# Handwritten Digit Recognition
Goal: recognize handwritten digits (0-9) from images using a Convolutional Neural Network (CNN).
Dataset: MNIST (built into Keras)

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras

(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()

print(X_train.shape, X_test.shape)
```

```python
plt.figure(figsize=(10, 2))
for i in range(10):
    plt.subplot(1, 10, i + 1)
    plt.imshow(X_train[i], cmap="gray")
    plt.title(y_train[i])
    plt.axis("off")
plt.show()
print(np.unique(y_train, return_counts=True))
```

Observations:
Training set: 60,000 images, Test set: 10,000 images.
Each image is a 28×28 grayscale image.
The data is fairly balanced with 10 (digits 0-9) classes.

```python

(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()

X_train = X_train / 255.0
X_test = X_test / 255.0

X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)

print(X_train.shape, X_test.shape)
print("Max pixel value:", X_train.max())
```

Data preparation:
- Values of pixels were normalized from 0-255 to 0-1.
Images were reshaped to (28, 28, 1) as it needs a channel dimension.
MNIST is already split into training and test sets so no additional split was necessary.

```python
from tensorflow.keras import layers


model = keras.Sequential([
    layers.Input(shape=(28, 28, 1)),

    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),


    layers.Dense(64, activation="relu"),
    layers.Dense(10, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
history = model.fit(
    X_train, y_train,
    epochs=5,
    batch_size=128,
    validation_split=0.1
)
```

```python
test_loss, test_acc = model.evaluate(X_test, y_test)
print("Test accuracy:", test_acc)
```

Model building and training:
I have two convolution + max pooling layers, a dense layer and a softmax output layer with 10 classes, all of which I built.
Adam was used as an optimizer for the model, which was trained for 5 epochs, with 10% of the training data being reserved for validation.
The CNN is also better than basic image models since it is able to learn edges and shapes directly from pixels.

Model building, training:
I had two layers of convolution and two layers of max pooling, followed by a dense layer, and a softmax output layer of 10 classes.
The model was trained for 10% of the training data using the Adam optimizer for 5 epochs.
The accuracy of the trained model increased from 92.5% in the first epoch to 99.0% in the final epoch with the accuracy of trained model validation as 99.0%.
The training and validation accuracy are close, indicating that the model is not overfitting.
The model achieved a test accuracy of 98.93% on the unseen test images of 10,000 images.

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history["accuracy"], label="Training")
plt.plot(history.history["val_accuracy"], label="Validation")
plt.title("Accuracy")
plt.xlabel("Epoch")
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history["loss"], label="Training")
plt.plot(history.history["val_loss"], label="Validation")
plt.title("Loss")
plt.xlabel("Epoch")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay


y_pred = np.argmax(model.predict(X_test), axis=1)

cmd = confusion_matrix(y_test, y_pred)
ConfusionMatrixDisplay(cmd).plot(cmap="Blues")
plt.title("CNN - Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
wrong = np.where(y_pred != y_test)[0]
print("Number of wrong predictions:", len(wrong))


plt.figure(figsize=(12, 2.5))
for i, idx in enumerate(wrong[:10]):
    plt.subplot(1, 10, i + 1)
    plt.imshow(X_test[idx].reshape(28, 28), cmap="gray")
    plt.title(f"True: {y_test[idx]}\nPred: {y_pred[idx]}", fontsize=8)
    plt.axis("off")
plt.show()
```

Evaluation:
The accuracy of the training curves and the loss are improving, and there is no overfitting as the training and the validation curves are close.
Most of the digits are correctly classified, as indicated by large diagonal values in the confusion matrix.
Most errors occur when the digits are similar when handwritten, like 4 and 9.
- Final accuracy: 98.93% on the 10000 test images that were not seen during training.
